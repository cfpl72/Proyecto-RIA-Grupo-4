```python
import os
import shutil

source_dir = r"C:\Users\music\Downloads\archive\fruits-360_original-size\fruits-360-original-size\Training"
target_dir = "dataset_final"

for folder in os.listdir(source_dir):

    folder_path = os.path.join(source_dir, folder)

    if "Apple" in folder:
        target_class = "apple"
    elif "Tomato" in folder:
        target_class = "tomato"
    elif "Pepper" in folder:
        target_class = "pepper"
    else:
        continue

    for img in os.listdir(folder_path):
        src = os.path.join(folder_path, img)
        dst = os.path.join(target_dir, target_class, img)

        shutil.copy(src, dst)
        print(folder)
```

```python
import tensorflow as tf

img_size = (224, 224)
batch_size = 32

dataset = tf.keras.preprocessing.image_dataset_from_directory(
    "dataset_final",
    image_size=img_size,
    batch_size=batch_size,
    shuffle=True
)

class_names = dataset.class_names
print(class_names)
```

```python
normalization_layer = tf.keras.layers.Rescaling(1./255)

dataset = dataset.map(lambda x, y: (normalization_layer(x), y))
```

```python
data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip("horizontal"),
    tf.keras.layers.RandomRotation(0.1),
    tf.keras.layers.RandomZoom(0.1),
    tf.keras.layers.RandomBrightness(0.1)
])
dataset = dataset.map(lambda x, y: (data_augmentation(x), y))
```

```python
train_size = int(0.8 * len(dataset))

train_dataset = dataset.take(train_size)
val_dataset = dataset.skip(train_size)
```
